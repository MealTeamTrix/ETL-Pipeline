```python
import pandas as pd
import json
```

```python
def extract_tabular_data(file_path: str) -> pd.DataFrame:
    """
    Extract data from a CSV or Parquet file.

    Parameters:
        file_path (str): Path to the file.

    Returns:
        pd.DataFrame: Loaded data.

    Raises:
        ValueError: If the file extension is not supported.
    """
    if file_path.endswith(".csv"):
        return pd.read_csv(file_path)
    elif file_path.endswith(".parquet"):
        return pd.read_parquet(file_path)
    else:
        raise ValueError("Invalid file extension. Please use .csv or .parquet.")
```

```python
def extract_json_data(file_path: str) -> pd.DataFrame:
    """
    Extract and flatten data from a JSON file.

    Parameters:
        file_path (str): Path to the JSON file.

    Returns:
        pd.DataFrame: Flattened JSON data.
    """
    with open(file_path, "r") as file:
        raw_data = json.load(file)
    return pd.json_normalize(raw_data)
```

```python
def transform_electricity_sales_data(df: pd.DataFrame) -> pd.DataFrame:
    """
    Transform electricity sales data.

    Parameters:
        df (pd.DataFrame): Raw electricity sales data.

    Returns:
        pd.DataFrame: Transformed data with selected sectors and fields.
    """
    df = df.dropna(subset=["price"])
    df = df[df["sectorName"].isin(["residential", "transportation"])].copy()

    df["year"] = df["period"].str[:4]
    df["month"] = df["period"].str[5:]

    return df[["year", "month", "stateid", "price", "price-units"]]
```

```python
def load_data(df: pd.DataFrame, file_path: str) -> None:
    """
    Save DataFrame to a CSV or Parquet file.

    Parameters:
        df (pd.DataFrame): Data to save.
        file_path (str): Destination file path.

    Raises:
        ValueError: If the file extension is not supported.
    """
    if file_path.endswith(".csv"):
        df.to_csv(file_path, index=False)
    elif file_path.endswith(".parquet"):
        df.to_parquet(file_path, index=False)
    else:
        raise ValueError(f"{file_path} is not a valid file type. Use .csv or .parquet.")
```

```python
from data_pipeline import (
    extract_tabular_data,
    extract_json_data,
    transform_electricity_sales_data,
    load_data
)

# Example: Extract from CSV and transform
df = extract_tabular_data("electricity_sales.csv")
transformed = transform_electricity_sales_data(df)
load_data(transformed, "cleaned_sales.parquet")

# Example: Extract from JSON
json_df = extract_json_data("data.json")
```
